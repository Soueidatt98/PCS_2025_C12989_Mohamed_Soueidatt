### Ex. 1 - Test de similarite par les valeurs propres

```python
import numpy as np

def have_same_eigenvalues(A, B, tol=1e-10):
    eig_A = np.sort_complex(np.linalg.eigvals(A))
    eig_B = np.sort_complex(np.linalg.eigvals(B))
    return np.allclose(eig_A, eig_B, atol=tol)

A = np.array([[2., 1.], [0., 3.]])
S = np.array([[1., 2.], [3., 5.]])
B = np.linalg.inv(S) @ A @ S
have_same_eigenvalues(A, B), 'functional test: it checks a mathematical property of the whole operation' 
```

### Ex. 2 - Comparaison du temps pour le produit scalaire

```python
import timeit
v = np.random.default_rng(0).random(100_000)
w = np.random.default_rng(1).random(100_000)

t_np = timeit.timeit('v @ w', number=50, globals=globals())
t_gen = timeit.timeit('sum((x*y for x, y in zip(v, w)))', number=10, globals=globals())
t_list = timeit.timeit('sum([x*y for x, y in zip(v, w)])', number=10, globals=globals())
{'numpy_seconds': round(t_np, 4), 'generator_seconds': round(t_gen, 4), 'list_seconds': round(t_list, 4)}
```

### Ex. 3 - Moyenne mobile: vectorisation contre comprehension

```python
u = np.random.default_rng(2).random(1_000_000)

t_vec = timeit.timeit('(u[:-2] + u[1:-1] + u[2:]) / 3', number=20, globals=globals())
t_list = timeit.timeit('np.array([(u[i] + u[i + 1] + u[i + 2]) / 3 for i in range(len(u)-3)])', number=1, globals=globals())
round(t_vec, 4), round(t_list, 4), 'the vectorized NumPy expression is faster' 
```
