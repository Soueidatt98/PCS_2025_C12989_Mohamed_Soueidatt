### Ex. 2 - Generateur de la suite `u_n = 2 u_{n-1}`

```python
import itertools

def geometric_sequence(u0):
    u = u0
    while True:
        yield u
        u = 2*u

list(itertools.islice(geometric_sequence(3), 8))
```

```python
def even_numbers():
    n = 0
    while True:
        yield n
        n += 2

list(itertools.islice(even_numbers(), 10))
```

### Ex. 4 - Convergence de `s_n = (1 + 1/n)^(2n)` vers `e^2`

```python
def s_values():
    n = 1
    while True:
        yield n, (1 + 1/n)**(2*n)
        n += 1

for n, s in s_values():
    if abs(s - math.e**2) < 1e-5:
        answer = (n, s)
        break
answer
```

### Ex. 5 - Crible d'Eratosthene

```python
def primes_less_than(n):
    if n <= 2:
        return []
    is_prime = [True]*n
    is_prime[0:2] = [False, False]
    for p in range(2, int(math.sqrt(n)) + 1):
        if is_prime[p]:
            for k in range(p*p, n, p):
                is_prime[k] = False
    return [p for p in range(n) if is_prime[p]]

primes_less_than(50)
```

### Ex. 6 - Euler explicite pour `u' = -sin(u)`

```python
def euler_sin(u0, h):
    u = u0
    while True:
        yield u
        u = u - h*math.sin(u)

[round(u, 6) for u in itertools.islice(euler_sin(1.0, 0.1), 8)]
```

### Ex. 7 - Approximation de pi par la regle des trapezes

```python
def trapezoid_values(f, a, b, n):
    h = (b - a)/n
    for i in range(n + 1):
        yield f(a + i*h)

f = lambda x: 4/(1 + x*x)
n = 10_000
h = 1/n
ys = trapezoid_values(f, 0, 1, n)
first = next(ys)
middle_sum = 0
last = None
for y in ys:
    if last is not None:
        middle_sum += last
    last = y
pi_approx = h*(0.5*first + middle_sum + 0.5*last)
pi_approx, abs(pi_approx - math.pi)
```

### Ex. 9 - Nombre d'iterations de l'AGM

```python
def agm_iterations(m, tol=1e-12):
    a, b = 1.0, math.sqrt(1 - m)
    count = 0
    while abs(a - b) > tol:
        a, b = (a + b)/2, math.sqrt(a*b)
        count += 1
    return count, math.pi/(2*a)

agm_iterations(0.5)
```

### Ex. 10 - Recursion pour `E_n = int_0^1 x^n exp(x-1) dx`

```python
def E_forward(nmax):
    E = math.exp(-1)
    yield 1, E
    for n in range(2, nmax + 1):
        E = 1 - n*E
        yield n, E

def E_backward(nmax):
    values = {nmax: 0.0}
    E = 0.0
    for n in range(nmax, 1, -1):
        E = (1 - E)/n
        values[n - 1] = E
    return values

forward20 = list(E_forward(20))[-1][1]
backward = E_backward(20)
round(forward20, 6), round(backward[1], 12), round(backward[10], 8)
```

### Ex. 11 - Produit d'Euler pour approximer `sin(x)`

```python
def euler_sine_product(k, x):
    p = x
    for j in range(1, k + 1):
        p *= 1 - x*x/(j*j*math.pi*math.pi)
    return p

xs = [-1 + (3.5*math.pi + 1)*i/199 for i in range(200)]
errors = {k: max(abs(euler_sine_product(k, x) - math.sin(x)) for x in xs)
          for k in [5, 10, 20, 50, 100]}
errors
```
